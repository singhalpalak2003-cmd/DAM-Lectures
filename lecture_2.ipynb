```python
#byjus case
import pandas as pd
import numpy as np

df = pd.read_excel("/content/Lecture2_data.xlsx")
```

```python
#data description
print(df.head)
print(df.shape)
df.info()
```

```python
#data type
print(df .dtypes)
```

```python
#top 10 profit/loss prodcut
df["Total_Cost_Lakh"] = (
    df["Marketing_Spend_Lakh"]
    + df["Content_Cost_Lakh"]
    + df["Salary_Cost_Lakh"]
    + df["Other_Cost_Lakh"]
)

df["Profit_Lakh"] = df["Revenue_Lakh"] - df["Total_Cost_Lakh"]

print(df[["Product", "Revenue_Lakh", "Total_Cost_Lakh", "Profit_Lakh"]].head(10))

```

```python
#Missing Value
print(df.isna().sum())
```

```python
#Decision making structures
#If Statements
rating = df["Customer_Rating"].mean()

if rating >= 4:
  print("Customer satisfaction is good")
```

```python
revenue = df["Revenue_Lakh"].mean()

if revenue > 400:
  print("Revenue is good")
else:
  print("Revenue is bad")
```

```python
revenue = df["Revenue_Lakh"].mean()
print("Average Revenue:", revenue)

if revenue > 300:
  print("Average revenue is high")
```
